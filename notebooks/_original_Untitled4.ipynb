```python
import pandas as pd

df = pd.read_csv('creditcard.csv')

mean_amount = df['Amount'].mean()
std_amount = df['Amount'].std()
df['amount_zscore'] = (df['Amount'] - mean_amount) / std_amount

df['hour'] = (df['Time'] / 3600) % 24
df['is_night'] = ((df['hour'] >= 0) & (df['hour'] <= 6)).astype(int)

print(df.shape)
print(df.columns.tolist())
```

```python
from sklearn.ensemble import IsolationForest

features = df[['amount_zscore', 'hour', 'Amount']]

model = IsolationForest(contamination=0.0017, random_state=42)
model.fit(features)

df['ml_flagged'] = model.predict(features)

print(df['ml_flagged'].value_counts())
```

```python
fraud_caught = df[(df['ml_flagged'] == -1) & (df['Class'] == 1)].shape[0]
total_flagged = df[df['ml_flagged'] == -1].shape[0]
total_fraud = df[df['Class'] == 1].shape[0]

recall = fraud_caught / total_fraud * 100
precision = fraud_caught / total_flagged * 100

print("Fraud caught:", fraud_caught, "out of", total_fraud)
print("Total flagged:", total_flagged)
print("Recall:", recall, "%")
print("Precision:", precision, "%")
```
